## Processing outputs
Because the scrapper can take very long time to run across years, it might be that we output the scrapped text into different files. With this notebook we can merge them and verify we have all the dates for the years we want to analyze.

```python
import pandas as pd

df1 = pd.read_csv('../output/guardian_headlines_full0.csv')

df2 = pd.read_csv('../output/guardian_headlines_full6.csv')
```

```python
# merge the two dataframes
df = pd.concat([df1, df2])
```

```python
# convert column date to datetime
df['date'] = pd.to_datetime(df['date'])

# group by year and count number of unique values in column date
df['year'] = df['date'].dt.year
df_grouped = df.groupby('year').agg({'date': 'nunique'}).reset_index()
```

```python
# Save the dataframe to a csv file
df.to_csv('../output/guardian_headlines_all.csv', index=False)
```
